```python
from data_loaders import get_dataset
from data_utils import split_data
import dill as pickle
import numpy as np
import matplotlib.pyplot as plt
```

```python
def load_cond_density_estimator(file_path):
    with open(file_path, 'rb') as f:
        cond_density_estimator = pickle.load(f)
    return cond_density_estimator

def calculate_fontsize(figsize):
    # Calculate fontsize based on figsize
    return int(5 * (figsize[0] / 10))

def plot_density_and_histograms(X, y, r, features, d, cond_density_estimator, variable_type, country):
    feature_matrix = np.eye(d)
    label_matrix = features[:d]
    densities = cond_density_estimator(feature_matrix)


    bins = np.linspace(min(y), np.quantile(y, 0.99), 100)
    colors = ["tab:blue", "tab:orange", "tab:green", "tab:red", "tab:purple"]

    figsize=(d * 10, d+3)
    fig, ax = plt.subplots(1, d, figsize=figsize)
    fontsize = calculate_fontsize(figsize)  # Increase font size for better readability

    for i in range(d):
        region = label_matrix[i].split("_")[-1]
        ax[i].plot(bins, densities[i].pdf(bins), label="Estimated Density " + r"$\hat{f}(y \mid x)$", color="black")
        ax[i].hist(y[X[:, i] == 1.], label=label_matrix[i], bins=bins, color=colors[i], weights=r[X[:, i] == 1.], density=True)
        ax[i].set_title("{} {}".format(region, variable_type), fontsize=fontsize)
        ax[i].set_xlabel("Daily Expenditures per Capita " + r"$y$", fontsize=fontsize)  # Increase font size
        ax[i].set_ylabel("Density", fontsize=fontsize)  # Increase font size
        ax[i].tick_params(axis='both', labelsize=fontsize-2)  # Increase font size for ticks

    plt.suptitle("{} Daily Expenditures per Capita Conditional on {}".format(country.capitalize(), variable_type), fontsize=fontsize+2)  # Increase font size
    #plt.savefig("figs/{}_consumption_by_{}.pdf".format(country, variable_type.lower()), bbox_inches="tight")
    plt.show()
```

```python
# Load dataset
X, y, r, features = get_dataset("malawi")
d = 3
(X_train, y_train, r_train), (X_test, y_test, r_test) = split_data(X=X, y=y, r=r, p=0.6)

# Load the pickled conditional density estimator
cond_density_estimator = load_cond_density_estimator('malawi_example_cond_density_estimator.pickle'.format(d))

# Plot density and histograms
plot_density_and_histograms(X_train, y_train, r_train, 
                            features, d, cond_density_estimator, "Dwelling Type", "malawi")
```
