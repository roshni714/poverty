```python
from opt_targeted_transfers import OracleGapTargetedTransfers, OracleRateTargetedTransfers
from opt_targeted_transfers import Dataset
from data_loaders import load_data, PATH_TO_TEST_DATA
```

```python
test_data = load_data(PATH_TO_TEST_DATA)
test_dataset = Dataset(df=test_data, outcome='consumption_per_capita_per_day', weight='hh_wgt', covs=['hh_size', 'urban'])
```

```python
tt = OracleRateTargetedTransfers(c_bar=2.15, budget=0.25)
```

```python
# Run optimization algorithm
assignments = tt.run_opt(test_dataset)
```

```python
# Evaluate policy. 
res = tt.evaluate(test_dataset)
res
```

```python
tt.compute_auc(test_dataset=test_dataset, 
               metrics=['post_transfer_poverty_rate', 'post_transfer_poverty_gap'], 
               budgets=[0.1, 0.25, 0.5, 0.75, 1.0, 1.5, 2.0, 2.15])
```

```python
tt = OracleGapTargetedTransfers(c_bar=2.15, budget=0.25, scheme="lift_to_line")
tt.run_opt(test_dataset)
res = tt.evaluate(test_dataset)
res #lift to line scheme behaves the same as the oracle poverty rate method -- allocates transfers to units closest to poverty line
```

```python
tt.compute_auc(test_dataset=test_dataset, 
               metrics=['post_transfer_poverty_rate', 'post_transfer_poverty_gap'], 
               budgets=[0.1, 0.25, 0.5, 0.75, 1.0, 1.5, 2.0, 2.15])
```

```python
tt = OracleGapTargetedTransfers(c_bar=2.15, budget=0.25, scheme="floor")
tt.run_opt(test_dataset)
res = tt.evaluate(test_dataset)
res #floor scheme allocates transfers to units furthest from the poverty line-- as a result, does not yield a large change in poverty rate for small budget
```

```python
tt.compute_auc(test_dataset=test_dataset, 
               metrics=['post_transfer_poverty_rate', 'post_transfer_poverty_gap'], 
               budgets=[0.1, 0.25, 0.5, 0.75, 1.0, 1.5, 2.0, 2.15])
```
