```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def plot_policy_costs(df, unconditional_tolerance, district):
    # Filter rows based on tolerance
    filtered_df = df[df['unconditional_tolerance'] == unconditional_tolerance]

    policy_order = {'saturation': 0, 'geographic': 1, 'binary_targeted': 2, 'optimized': 3, 'oracle': 4}

    # Sort filtered DataFrame based on policy order
    sorted_df = filtered_df.set_index('policy').loc[policy_order.keys()].reset_index()


    # Extract policy names and costs
    policies = sorted_df['policy'].tolist()
    policy_costs = sorted_df['policy_cost'].tolist()

    # Assign different colors to each policy
    colors = plt.cm.get_cmap('tab20', len(policies))

    # Create bar graph with different colors
    plt.figure(figsize=(8, 6))
    bars = plt.bar(policies, policy_costs, color=colors(range(len(policies))))

    # Add labels and title
    plt.xlabel('Policy')
    plt.ylabel('Policy Cost')
    plt.title(f'{district} (Tolerance: {unconditional_tolerance})')

    # Rotate x-axis labels for better visibility
    plt.xticks(rotation=45, ha='right')

    # Add legend
    plt.legend(bars, policies, title='Policy')

    # Display the plot
    plt.tight_layout()
    plt.savefig("figs/{}.pdf".format(district))
    plt.show()
    plt.close()
```

```python
# Example usage
districts = ["dowa", "kasungu", "mchinji"]
tolerances = [0.05, 0.10, 0.20, 0.30, 0.40]

for tolerance in tolerances:
    for district in districts:
        df = pd.read_csv('results/{}.csv'.format(district))
        plot_policy_costs(df, unconditional_tolerance=tolerance, district=district)  # Change tolerance value as needed
```
